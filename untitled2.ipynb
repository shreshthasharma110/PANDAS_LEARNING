Creating a series using dictionary

```python
fruit_protien ={
    'Avocado': 2.0 ,  # protien in grams
    'Apple':5.0,
    'Blackberries':1.1,
    'Kiwi':7.0,
    'Mango':3.3,
    'Cherries':4.2
}
```

```python
s2 = pd.Series(fruit_protien , name = 'protien in gm')
s2
```

```python
# Logical Operator : and, or, not
[(s2>1.1) & (s2<5.5 )]
```

```python
# printing only those values that satisfying this condition
s2[(s2>1.1) & (s2<5.5 )]
```
