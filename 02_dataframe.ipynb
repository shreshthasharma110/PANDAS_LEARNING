DATAFRAME


```python
import pandas as pd
import numpy as np
```

```python
data = {'Name':['Alice','bob','charlie','david','Eve'],
        'Age':[25,30,np.nan,28,35],
        'Department':['HR','IT','Finance','IT',"HR"],
        'Salary':[50000,60000,70000,np.nan,50000],
        }
```

```python
data
```

```python
df = pd.DataFrame(data)
df
```

```python
# loc and iloc function to fetch
# iloc for position based indexing
# loc for label based indexing

df.iloc[1:3]
```

```python
df.loc[1:3 ,['Age','Department']]
```
